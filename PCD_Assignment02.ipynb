Blur Image

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from google.colab import drive

drive.mount('/content/drive', force_remount=True)

image_path = '/content/drive/MyDrive/Semester Uni Folders/Semester 3/Enhancement Images/blur.jpeg'
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

linear = np.array([[-2, -1,  0],
                   [-1,  1,  1],
                   [ 0,  1,  2]])
img_filter = cv2.filter2D(src=img, ddepth=-1, kernel=linear)
img_enhance = cv2.equalizeHist(img_filter)

m, n = img.shape
print('Image Shape:', m, n)

images = [img, img_filter, img_enhance]
titles = ['Original', 'Sharpened Filter', 'Histogram Equalization']

plt.figure(figsize=(12, 4))
for i in range(3):
    plt.subplot(1, 3, i + 1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

High Brightness

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from google.colab import drive

drive.mount('/content/drive', force_remount=True)

#image_path = '/content/drive/MyDrive/Semester Uni Folders/Semester 3/Enhancement Images/bright.jpg'
image_path = '/content/drive/MyDrive/Semester Uni Folders/Semester 3/Enhancement Images/bright2.jpg'
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

gamma = 4.0
gamma_corrected = np.power(img / 255.0, gamma) * 255.0
gamma_corrected = gamma_corrected.astype(np.uint8)

m, n = img.shape
print('Image Shape:', m, n)

images = [img, gamma_corrected]
titles = ['Original', 'Gamma Correction']

plt.figure(figsize=(12, 4))
for i in range(2):
    plt.subplot(1, 2, i + 1)
    plt.imshow(images[i], cmap='gray', vmin=0, vmax=255)
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

Dark Image

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from google.colab import drive

drive.mount('/content/drive', force_remount=True)
#image_path= '/content/drive/MyDrive/Semester Uni Folders/Semester 3/Enhancement Images/dark.jpeg'
#image_path= '/content/drive/MyDrive/Semester Uni Folders/Semester 3/Enhancement Images/dark2.jpeg'
image_path= '/content/drive/MyDrive/Semester Uni Folders/Semester 3/Enhancement Images/dark3.jpeg'
#image_path= '/content/drive/MyDrive/Semester Uni Folders/Semester 3/Enhancement Images/dark4.jpeg'

img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
img_clahe = clahe.apply(img)

m, n = img.shape
print('Image Shape:', m, n)

images = [img, img_clahe]
titles = ['Original', 'CLAHE']


plt.figure(figsize=(12, 4))
for i in range(2):
    plt.subplot(1, 2, i + 1)
    plt.imshow(images[i], cmap='gray', vmin=0, vmax=255)
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

Low Contrast

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from google.colab import drive

drive.mount('/content/drive', force_remount=True)

image_path = '/content/drive/MyDrive/Semester Uni Folders/Semester 3/Enhancement Images/low contrast.jpeg'
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

p5, p95 = np.percentile(img, (5, 95))
stretched = np.clip(((img - p5) / (p95 - p5)) * 255, 0, 255).astype(np.uint8)

m, n = img.shape
print('Image Shape:', m, n)

images = [img, stretched]
titles = ['Original Low Contrast', 'Stretching (5%-95% Clip)']

plt.figure(figsize=(12, 4))
for i in range(2):
    plt.subplot(1, 2, i + 1)
    plt.imshow(images[i], cmap='gray', vmin=0, vmax=255)
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
